```python
original_train_y['SalePrice'].shape
original_train_y['SalePrice'].describe()
```

```python
quantiles = original_train_y['SalePrice'].quantile([0, 0.25, 0.5, 0.75, 1.0])
quantile_bins = pd.qcut(original_train_y['SalePrice'], q=4)
print(quantile_bins.value_counts())
```

```python
labels = ['Low', 'Medium', 'High', 'Luxury']

original_train_y['PriceCategory'] = pd.cut(
    original_train_y['SalePrice'],
    bins=quantiles,
    labels=labels,
    include_lowest=True
)

print(original_train_y['PriceCategory'].value_counts())
```
