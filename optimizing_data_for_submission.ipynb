### Handling Missing Data

test

```python
# 1. Fill absence indicators 
none_cols = ['PoolQC','MiscFeature','Alley','Fence','FireplaceQu',
             'GarageType','GarageFinish','GarageQual','GarageCond',
             'BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1',
             'BsmtFinType2','MasVnrType']
test_dataset_for_submission[none_cols] = test_dataset_for_submission[none_cols].fillna('None')

# 2. LotFrontage: fill missing values with median by neighborhood
lotfrontage_median = test_dataset_for_submission.groupby('Neighborhood')['LotFrontage'].transform('median')
test_dataset_for_submission['LotFrontage'] = test_dataset_for_submission['LotFrontage'].fillna(lotfrontage_median)

# 3. GarageYrBlt: fill with median year
test_dataset_for_submission['GarageYrBlt'] = test_dataset_for_submission['GarageYrBlt'].fillna(test_dataset_for_submission['GarageYrBlt'].median())

# 4. MasVnrArea: fill with 0
test_dataset_for_submission['MasVnrArea'] = test_dataset_for_submission['MasVnrArea'].fillna(0)

# 5. Electrical: fill with mode
test_dataset_for_submission['Electrical'] = test_dataset_for_submission['Electrical'].fillna(test_dataset_for_submission['Electrical'].mode()[0])
```

train

```python
# 1. Fill absence indicators 
none_cols = ['PoolQC','MiscFeature','Alley','Fence','FireplaceQu',
             'GarageType','GarageFinish','GarageQual','GarageCond',
             'BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1',
             'BsmtFinType2','MasVnrType']
train_dataset_with_target[none_cols] = train_dataset_with_target[none_cols].fillna('None')

# 2. LotFrontage: fill missing values with median by neighborhood
lotfrontage_median = train_dataset_with_target.groupby('Neighborhood')['LotFrontage'].transform('median')
train_dataset_with_target['LotFrontage'] = train_dataset_with_target['LotFrontage'].fillna(lotfrontage_median)

# 3. GarageYrBlt: fill with median year
train_dataset_with_target['GarageYrBlt'] = train_dataset_with_target['GarageYrBlt'].fillna(train_dataset_with_target['GarageYrBlt'].median())

# 4. MasVnrArea: fill with 0
train_dataset_with_target['MasVnrArea'] = train_dataset_with_target['MasVnrArea'].fillna(0)

# 5. Electrical: fill with mode
train_dataset_with_target['Electrical'] = train_dataset_with_target['Electrical'].fillna(train_dataset_with_target['Electrical'].mode()[0])
```

- In order to don't miss any value, I decided to fill it with __"None"__ for the houses that doesn't have some housing features details, such as basement, pool, garage and so on.
- Also for the __"Linear feet of street connected to property"__ and __Year garage was built__,  miss values, I filled it with the __median__
- For the __Masonry veneer area in square feet__ The best option, based on the column describe, is to fill the missing values with "0". (Most of the houses doesn't have masonry veneer).
- The __"Eletrical feature"__ is categorical, so based on it, I used the __"mode"__ to fill the missing data.

| `TOTAL_ROOMS`| `TotRmsAbvGrd`;`FullBath`;`HalfBath`;`BsmtFullBath`; `BsmtHalfBath`| Total rooms above grade + Full bathrooms and Half above grade + Basement full bathrooms and half bath|
|`TOTAL_AREA`|`GrLivArea`;`TotalBsmtSF`| Above ground living area square feet + Total square feet of basement area|
| `TOTAL_VOL`|`MasVnrArea`; `GrLivArea`|  Exterior veneer with the size of the living area, scaled down|

- In our dataset, we experimented with feature engineering by creating new variables. However, the results were not successful, instead of improving model performance, the additional features actually made it **worse**. We suspected this was due to redundance, as the new features were essentially derived from existing ones and provided overlapping information. To test this, we dropped the original columns used to create those features, but even then the performance did not improve. Given these outcomes, we decided to set aside feature engineering for now and continue working with the dataset in its original form.
